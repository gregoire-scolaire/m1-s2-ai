### Classe Futoshiki
Futoshiki est différent du taquin.
 - Il n'y a pas d'état final unique mais plusieurs états finaux qui répondent au test de réussite.
 - Les actions ne sont plus (haut bas gauche droite) mais un dictionnaire de position (tuple) et liste des nombres qui peuvent être mis dessus (pour BFS de base, nombre de 1-5).
 - Les états initiaux possibles sont des grilles déjà pré-rempli à certains endroits, comme pour le sudoku vu en TD.
 - La fonction action renvoie toutes les actions disponibles, **y compris** celles qui devraient etre impossibles vis à vis des règles du Futoshiki.
 - La classe de Futoshiki a besoin de deux éléments : la grille contenant les chiffres, et un dictionnaire permettant de situer les signes ">" et "<" entre les éléments cette grille.

```python

from IPython.display import display, HTML

def visualize_state(state, futoshi):
    #state = futoshi.initial_board
    constraint = futoshi.board_sup_inequality
    finalArray = []
    # Preprocessing les états pour avoir un seul tableau en HTML
    for row in state:
        oneLine = []
        #nextLine = []
        for tile in row:
            oneLine.append(tile)
            oneLine.append(".")
            #nextLine.append(".")
           # nextLine.append("")
        oneLine.pop(-1)
        #nextLine.pop(-1)
        finalArray.append(oneLine)
        #finalArray.append(nextLine)

    # Maintenant les constraints
    for (x1,y1),(x2,y2) in constraint:
        x,y = min(x1,x2), min(y1,y2)
        finalArray[x][2*y+1] = ">" if (x2>=x1 and y2>=y1) else "<"

    """Visualizes the given state of the Fu using HTML."""
    html = "<table>"
    for row in finalArray:
        html += "<tr>"
        for tile in row:
            if tile == 0:
                html += "<td style='background-color: #3a3a3a; width: 30px; height: 30px; text-align: center; font-size: 20px;'> </td>"  # Blank tile
            else:
                html += f"<td style='background-color: #3a3a3a; width: 30px; height: 30px; text-align: center; font-size: 20px;'>{tile}</td>"
        html += "</tr>"
    html += "</table>"
    display(HTML(html))
```

```python
class Futoshiki:
    """
    A class representing the futo problem.
    initial_board : le plateau de base, vide ou avec des chiffres
    board_inequality : les signes > et < du plateau. Ils sont mis sous la forme
    de tuple [((x1,y1), (x2,y2)), ( (x3,y3), (x4,y4) )] signifie par exemple que
    le chiffre en x1,y1 doit etre supérieur a x2,Y2:
    et que le chiffre en x3,y3 doit etre supérieur a celui en x4,y4
    """

    def __init__(self, initial_board, board_inequality):
        self.initial_state = initial_board
        self.board_sup_inequality = board_inequality
        self.size = len(initial_board)


    def actions(self, state):
         """
         Returns the possible actions for the FIRST empty cell found.
         This drastically reduces the search space size.
         """
         dicPlaceNumberPossible = {}

         # Scan the board
         for i in range(self.size):
             for j in range(self.size):
                 if state[i][j] == 0:
                     # We found the first empty cell!
                     # We define actions ONLY for this cell and return immediately.

                     # Note: We propose numbers 1 to Size.
                     # The checking (is_valid_move) will happen in the BFS.
                     possible_numbers = [x for x in range(1, self.size + 1)]

                     dicPlaceNumberPossible[(i, j)] = possible_numbers
                     return dicPlaceNumberPossible

         # If we get here, there were no zeros (board is full)
         return dicPlaceNumberPossible


    # Une seul action est une entrée du dictionnaire, place et nombre. L'entrée est converti en tuple
    # action = ( (x,y), chiffre). Par exemple, ( (0,0), 4) met le chiffre 4 à la position (0,0)
    def result(self, state, action):
        """Returns the state that results from applying the given action."""
        # Create a copy of the state to avoid modifying the original
        new_state = [list(row) for row in state]
        (x, y), chiffre = action
        new_state[x][y] = chiffre

        return new_state

    def is_goal(self, state):

        # dictionnaire clef : la ligne ou la colonne, set des chiffres de lignes et colonne.
        lines = {x: set() for x in range(5)}
        cols = {x: set() for x in range(5)}

        # Ajout de la valeur des lignes
        for i in range(self.size):
            for j in range(self.size):
                if state[i][j] == 0:
                    # LE NON SENS EST ICI JEAN GUY, ICI
                    pass
                    #return False

                # if( state[i][j] != 0):
                lines[i].add(state[i][j])
                cols[j].add(state[i][j])


        # Vérification des lignes & cols
        for key in lines:
            if len(lines[key]) != self.size or len(cols[key]) != self.size:
                return False
        # vérification des contraintes
        for inegality in self.board_sup_inequality:
            # ici, on a un tuple de tuple
            (x1, y1), (x2, y2) = inegality
            if state[x1][y1] < state[x2][y2]:
                return False
        return True

    def cost(self, state, action):
        return 1  # Default cost is 1
```

## Documentation python-constraint

The Python constraint module offers solvers for Constraint Satisfaction Problems (CSPs) over finite domains in simple and pure Python. CSP is class of problems which may be represented in terms of variables (a, b, …), domains (a in [1, 2, 3], …), and constraints (a < b, …).

### Example Basics  
This interactive Python session demonstrates the module basic operation:

```python
from constraint import *

problem = Problem()
problem.addVariable("a", [1,2,3])
problem.addVariable("b", [4,5,6])
problem.getSolutions()
#[{'a': 3, 'b': 6}, {'a': 3, 'b': 5}, {'a': 3, 'b': 4}, // Resultat du print
# {'a': 2, 'b': 6}, {'a': 2, 'b': 5}, {'a': 2, 'b': 4},
# {'a': 1, 'b': 6}, {'a': 1, 'b': 5}, {'a': 1, 'b': 4}]

problem.addConstraint(lambda a, b: a*2 == b,
                          ("a", "b"))
problem.getSolutions()
#[{'a': 3, 'b': 6}, {'a': 2, 'b': 4}]

problem = Problem()
problem.addVariables(["a", "b"], [1, 2, 3])
problem.addConstraint(AllDifferentConstraint())
problem.getSolutions()
#[{'a': 3, 'b': 2}, {'a': 3, 'b': 1}, {'a': 2, 'b': 3}, {'a': 2, 'b': 1}, {'a': 1, 'b': 2}, {'a': 1, 'b': 3}]
```

## classe Node

```python
class Node:
    """
    A node in a search tree.
    __init__: Initializes a node with its state, parent, action, path cost, and depth.
    __repr__: Provides a string representation of the node.
    __lt__: Defines a comparison operator for nodes based on their states.
    expand: Generates child nodes by applying all possible actions.
    child_node: Creates a single child node for a given action.
    solution: Returns the sequence of actions that led to this node.
    path: Returns the path from the root to this node as a list of nodes.
    """

    def __init__(self, state, parent=None, action=None, path_cost = 0):
        self.state = state
        self.parent = parent
        self.action = action
        self.path_cost = path_cost
        self.depth = 0 if parent is None else parent.depth + 1

    # def __repr__(self):
    #     return "<Node {}>".format(self.state)

    # def __lt__(self, other):
    #     return self.state < other.state

    def expand(self, problem):
        """List the nodes reachable in one step from this node."""
        return [
            self.child_node(problem, action)
            for action in problem.actions(self.state)
        ]

    def child_node(self, problem, action):
        """Create a child node by applying the given action."""
        next_state = problem.result(self.state, action)
        next_node = Node(
            next_state,
            parent=self,
            action=action,
            path_cost = self.path_cost + problem.cost(self.state, action),
        )
        return next_node

    def solution(self):
        """Return the sequence of actions to go from the root to this node."""
        return [node.action for node in self.path()[1:]]

    def path(self):
        """Return a list of nodes forming the path from the root to this node."""
        node, path_back = self, []
        while node:
            path_back.append(node)
            node = node.parent
        return list(reversed(path_back))
```

## Correction csp Sudoku

```python
from constraint import *

def solve_sudoku_csp(grid):
    problem = Problem()
    # Define variables (each cell in the grid) with domains (1-9)
    for row in range(9):
        for col in range(9):
            if grid[row][col] == 0:
                problem.addVariable((row, col), range(1, 10))
            else:
                problem.addVariable((row, col), [grid[row][col]])

    # Define constraints
    # AllDifferent constraint for rows, columns, and 3x3 subgrids
    for i in range(9):
        problem.addConstraint(AllDifferentConstraint(), [(i, j) for j in range(9)])  # Rows
        problem.addConstraint(AllDifferentConstraint(), [(j, i) for j in range(9)])  # Columns

    for i in range(0, 9, 3):
        for j in range(0, 9, 3):
            problem.addConstraint(AllDifferentConstraint(),
                                  [(i + a, j + b) for a in range(3) for b in range(3)])  # Subgrids

    # Get solutions
    solutions = problem.getSolutions()
    if solutions:  # Check if solutions were found
        # Convert the first solution (dictionary) to a 2D list
        first_solution = solutions[0]
        result_grid = [[0 for _ in range(9)] for _ in range(9)]
        for (row, col), value in first_solution.items():
            result_grid[row][col] = value
        return result_grid
    else:
        return None  # Or raise an exception if no solutions are found
```
